```python
from gurulearn import QAAgent
```

```python
import pandas as pd
data=pd.read_csv("TestReviews.csv")
model=QAAgent(
    data=data,
    llm_model="llama3.2",
    page_content_fields=["review"],
    metadata_fields=["class"],
    system_prompt="you are a helpful sentiment analysis assistant that classifies reviews and give better output for the customers"
    
)
```

```python
model.query("whats the best shop there")
```
